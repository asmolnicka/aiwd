<h3 style="color: dodgerblue;">Wybór podzbiorów poprzez numery kolumn i wierszy

Atrybut ramki danych `iloc` pozwala na indeksowanie wierszy i kolumn za pomocą ich numerów. \
Analogicznie jak w listach i numpy.array - numeracja zaczyna się od 0, a indeksowanie `a:b` oznacza przedział `[a, b)`.

**<p style="color: darkblue;">Jeden wiersz</p>**

```python
df_pracownicy.iloc[2]
```

**<p style="color: darkblue;">Kilka wierszy</p>**

```python
df_pracownicy.iloc[[2, 6]]
```

```python
df_pracownicy.iloc[2:7]
```

**<p style="color: darkblue;">Jedna komórka</p>**

```python
df_pracownicy.iloc[2, 0]
```

**<p style="color: darkblue;">Dowolny podzbiór</p>**

```python
df_pracownicy.iloc[[2, 6], 0]
```

```python
df_pracownicy.iloc[2, [0, 1]]
```

```python
df_pracownicy.iloc[[2, 6], [0, 1]]
```

```python
df_pracownicy.iloc[:6, 3:6]
```

<h3 style="color: dodgerblue;">Wybór podzbiorów poprzez nazwy kolumn i wierszy

Atrybut ramki danych `loc` pozwala na indeksowanie wierszy i kolumn za pomocą ich nazw. \
W przeciwieństwie do funkcji `iloc`, argument `'a':'b'` oznacza przedział `['a', 'b']`.

**<p style="color: darkblue;">Jeden wiersz</p>**

```python
df_pracownicy.loc['AN01']
```

**<p style="color: darkblue;">Kilka wierszy</p>**

```python
df_pracownicy.loc[['AN01', 'KM00']]
```

```python
df_pracownicy.loc['AN01':'KM00']
```

**<p style="color: darkblue;">Jedna komórka</p>**

```python
df_pracownicy.loc['AN01', 'Imię']                       
```

**<p style="color: darkblue;">Dowolny podzbiór</p>**

```python
df_pracownicy.loc[['AN01', 'KM00'], 'Imię']
```

```python
df_pracownicy.loc['AN01', ['Imię', 'Nazwisko']]  
```

```python
df_pracownicy.loc[['AN01', 'KM00'], ['Imię', 'Nazwisko']]
```

```python
df_pracownicy.loc[:'JM00', 'Zarobki':'Stanowisko']    
```

<h3 style="color: dodgerblue;">Wybór podzbiorów poprzez charakterystykę nazw kolumn i wierszy

Metoda ramki danych `filter` pozwala na zautomatyzowane indeksowanie wierszy i kolumn za pomocą wspólnej charakterystyki ich nazw.

**<p style="color: darkblue;">Wybieranie kolumn</p>**

```python
df_pracownicy.filter(['Imię', 'Nazwisko'])                # poprzez nazwy
```

```python
df_pracownicy.filter(like = 'ko')                         # poprzez obecność w nazwie podstringa
```

```python
df_pracownicy.filter(regex = 'sko$')                      # poprzez wyrażenie regularne (tu kończące się na "sko")
```

**<p style="color: darkblue;">Wybieranie wierszy</p>**

```python
df_pracownicy.filter(['AN01', 'KM00'], axis = 'rows')     # poprzez nazwy
```

```python
df_pracownicy.filter(like = 'M0',      axis = 'rows')     # poprzez obecność w nazwie podstringa
```

```python
df_pracownicy.filter(regex = '^A',     axis = 'rows')     # poprzez wyrażenie regularne (tu zaczynające się na "A")
```

Wszystkie z powyższych opcji odwołujących się do nazw kolumn lub wierszy, mogą zostać wykorzystane do zmiany ich kolejności, poprzez podanie wszystkich kolumn lub wierszy w ustalonym porządku.

**<h2 style="color: darkorange;">5. Operacje na ramkach danych</h2>**

Na pandas.dataframe można wykonywać podobny zbiór operacji jak na pandas.series, ale w większości przypadków bardziej praktyczne jest zastosowanie tych drugich - na kolumnach ramki danych.

<h3 style="color: dodgerblue;">Operacje matematyczne</h3>

```python
print(df_pracownicy['Zarobki'] + df_pracownicy['Premia'])                   # operacje matematyczne
print(df_pracownicy['Zarobki'] * 3)                                         # broadcasting
display(df_pracownicy[['Zarobki', 'Premia']] * 2)                           # broadcasting na DataFrame
```

<h3 style="color: dodgerblue;">Operacje stringowe</h3>

https://pandas.pydata.org/docs/reference/series.html#string-handling

```python
print(df_pracownicy['Nazwisko'] + '-' + df_pracownicy['Stanowisko'])        # konkatenacja stringów
print(df_pracownicy['Nazwisko'].str.lower())                                # małe litery
print(df_pracownicy['Nazwisko'].str.upper())                                # wielkie litery
print(df_pracownicy['Nazwisko'].str[0:3])                                   # podstring 
print(df_pracownicy['Nazwisko'].str.replace('ski', 'iak'))                  # zamiana podstringa
```

<h3 style="color: dodgerblue;">Operacje logiczne</h3>

**<p style="color: darkblue;">Z operatorami logicznymi</p>**

```python
print(df_pracownicy['Zarobki'] == (df_pracownicy['Premia'] + 3000))
print(df_pracownicy['Zarobki'] != (df_pracownicy['Premia'] + 3000))
print(df_pracownicy['Zarobki'] < (df_pracownicy['Premia'] + 3000))
print(df_pracownicy['Zarobki'] < 3100)
print((df_pracownicy['Zarobki'] > 3100) & (df_pracownicy['Zarobki'] < 3800))
```

**<p style="color: darkblue;">Z funkcjami o wartościach logicznych</p>**

```python
print(df_pracownicy['Stanowisko'].eq('sprzedawca'))                         # równość wartości
print(df_pracownicy['Zarobki'].isin([3100, 3800]))                          # należenie do zbioru
print(df_pracownicy['Imię'].isin(['Jan', 'Andrzej']))           
print(df_pracownicy['Nazwisko'].str.contains('ski'))                        # obecność podstringa
print(df_pracownicy['Zarobki'].isna())                                      # brak danych
```

<h3 style="color: dodgerblue;">Mapowanie</h3>

*Mapowanie danych* to przekształcanie poszczególnych wartości na inne wartości, zgodnie z podaną instrukcją

**<p style="color: darkblue;">Kodowanie:</p>**

```python
df_pracownicy['Stanowisko'].map({'programista' : 1,             
                                 'analityk'    : 2,
                                 'kierownik'   : 3,
                                 'sprzedawca'  : 4
                                })
```

Kodowanie jest często wykorzystywane do analizy zmiennych porządkowych.

**<p style="color: darkblue;">Przekształcenia liczbowe:</p>**

```python
print(df_pracownicy['Zarobki'] * 3)                                         # wcześniejszy sposób

def razy3(x):
    return x * 3

print(df_pracownicy['Zarobki'].map(razy3))                                  # argumentem map może być funkcja mapująca

print(df_pracownicy['Zarobki'].map(lambda x: x * 3))                        # funkcja "definiowana w miejscu", bez nazwy

print(df_pracownicy[['Zarobki', 'Premia']].map(lambda x: x * 3))            # map na DataFrame
```

**<p style="color: darkblue;">Klasyfikacja:</p>**

Na podstawie poszczególnych wartości:

```python
df_pracownicy['Stanowisko'].map({'programista' : 'IT',                
                                 'analityk'    : 'IT',
                                 'kierownik'   : 'MNG',
                                 'sprzedawca'  : 'BSN'
                                })
```

Na podstawie zbiorów wartości:

```python
print(df_pracownicy['Stanowisko'].map(lambda x: 'IT'  if x in ['programista', 'analityk'] else 'non-IT'))    
print(df_pracownicy['Stanowisko'].map(lambda x: 'IT'  if x in ['programista', 'analityk'] else (
                                                'MNG' if x == 'kierownik' else
                                                'BSN')))
```

Na podstawie przedziałów wartości - **<span style="color: darkblue;">polichotomizacja</span>** zmiennej:

```python
pd.qcut(df_pracownicy['Zarobki'],                             
        q      = 2,                                                         # liczba (jak najbardziej) równolicznych przedziałów
        labels = ['niskie', 'wysokie'])                                     # nazwy kolejnych klas 
```

```python
pd.cut(df_pracownicy['Zarobki'],                              
       bins   = [-float('inf'), 3200, float('inf')],                        # granice przedziałów: [-oo, 3200), [3200, +oo)
       right  = False,                                                      # czy przedziały lewostronnie domnknięte (domyślnie True)
       labels = ['niskie', 'wysokie'])                  
```

```python
pd.cut(df_pracownicy['Zarobki'],                                
       bins   = [0, 3000, 3300, float('inf')],                              # granice przedziałów: (0, 3000], (3000, 3300], (3300, +oo]
       labels = ['niskie', 'średnie', 'wysokie'])
```

Funkcje `qcut` i `cut` zwracają serię z odpowiednim typem danych - porządkowym z porządkiem kategorii zdefiniowanym przez argument `labels`.
