<h3 style="color: dodgerblue;">Zdefiniowanie typu serii</h3>

pandas.series zazwyczaj składa się z elementów jednego typu:

| typ serii                | elementy serii                                                | skala zmiennej           |
| -------------------------|---------------------------------------------------------------|--------------------------|
| `str`                    | ciągi znaków                                                  | nominalna                |
| `int64`                  | liczby całkowite                                              | ilościowa dyskretna      |
| `float64`                | liczby rzeczywiste                                            | ilościowa ciągła         |
| `bool`                   | wartości logiczne                                             | ilościowa zero-jedynkowa |
| `datetime64[ns]`         | daty i/lub czasy                                              | ilościowa czasowa        |
| `category` (ordered)     | kategorie określonego zbioru z ustalonym porządkiem           | porządkowa               |
| `category` (not ordered) | kategorie określonego zbioru                                  | nominalna                |
  
Do przechowywania w serii elementów różnego typu służy typ: `object`.

`64` w nazwach typów numerycznych oznacza dokładność liczby przechowywanej w pamięci komputera, tj. na ilu bitach jest ona zapisana. \
`ns` w typie danych dotyczących dat i czasów oznacza, że są one przechowywane w postaci nanosekund.

````python
zmienna1 = pd.Series([-1, 0, 1])                      # automatycznie dopasowany typ (prawdopodobnie str, int, float, bool lub object)
zmienna2 = pd.Series([-1, 0, 1], dtype = 'float')     
zmienna3 = pd.Series([-1, 0, 1], dtype = 'str')  

print(zmienna1)
print(zmienna2)
print(zmienna3)
print(zmienna4)
````

<h3 style="color: dodgerblue;">Operacje na seriach</h3>

Pandas.series posiadają podobny (ale nie identyczny) zbiór funkcjonalności do numpy.array:

````python
zmienna1 = pd.Series([1, 2, 3])
zmienna2 = pd.Series([4, 5, 6])

print(zmienna1 + zmienna2)                             # operacje matematyczne
print(zmienna1 * 2)                                    # broadcasting
print(pd.concat([zmienna1, zmienna2]))                 # konkatenacja
print(zmienna1[zmienna1 > 1])                          # operacje logiczne i filtrowanie
zmienna1[1] = 9; print(zmienna1)                       # modyfikacja
print(np.sqrt(zmienna1))                               # funkcje matematyczne
print(zmienna1.mean())                                 # funkcje statystyczne
````

````python
df_oryginal = pd.DataFrame({'a': [1, 4, 7], 'b': [2, 5, 8], 'c': ['pies', 'kot', 'krowa']})
display(df_oryginal)

df_2 = df_oryginal
df_3 = df_oryginal.loc[:, :]
df_4 = df_oryginal.copy()                              # funkcja kopiująca
````

**<h2 style="color: darkorange;">5. Operacje na ramkach danych</h2>**

Na pandas.dataframe można wykonywać podobny zbiór operacji jak na pandas.series, ale w większości przypadków bardziej praktyczne jest zastosowanie tych drugich - na kolumnach ramki danych.

<h3 style="color: dodgerblue;">Operacje matematyczne</h3>

````python
print(df_pracownicy['Zarobki'] + df_pracownicy['Premia'])                   # operacje matematyczne
print(df_pracownicy['Zarobki'] * 3)                                         # broadcasting
display(df_pracownicy[['Zarobki', 'Premia']] * 2)                           # broadcasting na DataFrame
````

<h3 style="color: dodgerblue;">Operacje stringowe</h3>

https://pandas.pydata.org/docs/reference/series.html#string-handling

````python
print(df_pracownicy['Nazwisko'] + '-' + df_pracownicy['Stanowisko'])        # konkatenacja stringów
print(df_pracownicy['Nazwisko'].str.lower())                                # małe litery
print(df_pracownicy['Nazwisko'].str.upper())                                # wielkie litery
print(df_pracownicy['Nazwisko'].str[0:3])                                   # podstring 
print(df_pracownicy['Nazwisko'].str.replace('ski', 'iak'))                  # zamiana podstringa
````

<h3 style="color: dodgerblue;">Operacje logiczne</h3>

**<p style="color: darkblue;">Z operatorami logicznymi</p>**

````python
print(df_pracownicy['Zarobki'] == (df_pracownicy['Premia'] + 3000))
print(df_pracownicy['Zarobki'] != (df_pracownicy['Premia'] + 3000))
print(df_pracownicy['Zarobki'] < (df_pracownicy['Premia'] + 3000))
print(df_pracownicy['Zarobki'] < 3100)
print((df_pracownicy['Zarobki'] > 3100) & (df_pracownicy['Zarobki'] < 3800))
````

**<p style="color: darkblue;">Z funkcjami o wartościach logicznych</p>**

````python
print(df_pracownicy['Stanowisko'].eq('sprzedawca'))                         # równość wartości
print(df_pracownicy['Zarobki'].isin([3100, 3800]))                          # należenie do zbioru
print(df_pracownicy['Imię'].isin(['Jan', 'Andrzej']))           
print(df_pracownicy['Nazwisko'].str.contains('ski'))                        # obecność podstringa
print(df_pracownicy['Zarobki'].isna())                                      # brak danych
````

<h3 style="color: dodgerblue;">Mapowanie</h3>

*Mapowanie danych* to przekształcanie poszczególnych wartości na inne wartości, zgodnie z podaną instrukcją

<h3 style="color: dodgerblue;">Konkatenacja</h3>

Przy dodawaniu wierszy podana została pozioma konkatenacja ramek. Ta sama funkcja, ale z argumentem `axis = 'columns'`, konkatenuje w poziomie, łacząc kolumny.

````python
#df_pracownicy =
pd.concat([df_pracownicy,                                                     
           pd.DataFrame({'A':np.random.randint(1, 10, 12),
                         'B':-99},
                         index = df_pracownicy.index)],
           axis = 'columns')    
````

Funkcja `concat` nie pozwala na przetwarzanie w tzw. potoku. W takiej sytuacji możemy skorzystać z funkcji `pipe` i podania w argumencie odpowiedniej funkcji.

````python
#df_pracownicy =
df_pracownicy.pipe(lambda df: pd.concat([df,
                                         pd.DataFrame({'A2':np.random.randint(1, 10, 12),
                                                       'B2':-99},
                                                      index = df_pracownicy.index)],
                                         axis = 'columns'))
````

**<h2 style="color: darkorange;">11. Przekształcenia ramki</h2>**

*długi format danych* - wartości zmiennej są w jednej kolumnie; każdy obserwacja ma swój wiersz

*szeroki format danych* - wartości zmiennej są w kilku kolumnach; wiersze mają po kilka obserwacji

````python
df_wide = plik_excel.parse('przypadki')
df_wide                                                    # format szeroki
````

<h3 style="color: dodgerblue;">Dane szerokie$ \rightarrow$ dane długie</h3>

````python
df_long = df_wide.melt(value_vars = [1999, 2000, 2001],    # zmienne, które mają być scalone
                       var_name   = 'Rok',                 # nazwa scalonej zmiennej
                       value_name = 'Przypadki',           # nazwa zmiennej z wartościami scalanych zmiennych
                       id_vars    = ['Kraj'])              # zmienne, które pozostawiamy       
df_long                                                    # format długi
````

<h3 style="color: dodgerblue;">Dane długie $\rightarrow$ dane szerokie</h3>

````python
df_wide_again = df_long.pivot(columns = 'Rok',             # zmienna rozdzielona na inne zmienne - jej wartości ich nazwami
                              values  = 'Przypadki',       # zmienna z rozdzielanymi wartościami
                              index   = 'Kraj')            # zmienna, która będzie indeksem
df_wide_again                                              # znów format szeroki
````
