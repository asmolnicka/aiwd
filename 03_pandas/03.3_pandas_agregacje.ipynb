```python
# załadowanie bibliotek

import pandas as pd
import numpy as np
```

```python
# wczytanie danych

df_mieszkancy = pd.ExcelFile('data_aggregation.xlsx').parse('aggregation_example')

# poprawa typów danych

df_mieszkancy = df_mieszkancy.assign(poziom_zaangazowania = df_mieszkancy['poziom_zaangazowania'].astype('category'
                                                                                                ).cat.set_categories(['I', 'II', 'III'], ordered = True),
                                     oplata_prad = df_mieszkancy['oplata_prad'].astype('float'))
```

```python
# podglad danych

display(df_mieszkancy)
df_mieszkancy.info()
```

<h1 style="color: red;">Opis statystyczny rozkładu zmiennej</h1>

**<h2 style="color: darkorange;">Miary położenia</h2>**

<h3 style="color: dodgerblue;">Miary tendencji centralnej</h3>

```python
print(df_mieszkancy['wiek'].mean())                             # średnia
print(df_mieszkancy['wiek'].median())                           # mediana
print(df_mieszkancy['wiek'].mode())                             # dominanta
```

Funkcja `mode` obsługuje również zmienne kategioralne, czyli nominalne lub porządkowe:

```python
print(df_mieszkancy['zawod'].mode())                            
print(df_mieszkancy['poziom_zaangazowania'].mode())
```

<h3 style="color: dodgerblue;">Inne miary położenia</h3>

```python
print(df_mieszkancy['wiek'].quantile(0.25))                     # kwantyl rzędu 0.25 (1. kwartyl)
print(df_mieszkancy['wiek'].quantile([0, 0.25, 0.5, 0.75, 1]))  # minimum, 1. kwartyl, 2.kwartyl, 3. kwartyl, maksimum
```

```python
print(df_mieszkancy['wiek'].min())                              # minimum  - kwantyl rzędu 0
print(df_mieszkancy['wiek'].median())                           # mediana  - kwantyl rzędu 0.5
print(df_mieszkancy['wiek'].max())                              # maksimum - kwantyl rzędu 1
```

Funkcje `quantile`, `min` i `max` obsługują również zmienne porządkowe:

```python
print(df_mieszkancy['poziom_zaangazowania'].quantile([0, 0.25, 0.5, 0.75, 1]))   
print(df_mieszkancy['poziom_zaangazowania'].min())
print(df_mieszkancy['poziom_zaangazowania'].quantile(0.5))
print(df_mieszkancy['poziom_zaangazowania'].max())
```

**<h2 style="color: darkorange;">Miary rozproszenia</h2>**

<h3 style="color: dodgerblue;">Miary rozproszenia wokół miary tendencji centralnej</h3>

```python
print(df_mieszkancy['wiek'].var())                              # wariancja nieobciążona
print(df_mieszkancy['wiek'].std())                              # odchylenie standardowe
```

<h3 style="color: dodgerblue;">Inne miary rozproszenia</h3>

```python
print(df_mieszkancy['zawod'].nunique())                         # liczba kategorii zmiennej kategorialnej
print(df_mieszkancy['poziom_zaangazowania'].nunique()) 
```

**<h2 style="color: darkorange;">Miary koncentracji</h2>**

```python
print(df_mieszkancy['wiek'].kurt())                             # kurtoza
```

**<h2 style="color: darkorange;">Miary asymetrii</h2>**

```python
print(df_mieszkancy['wiek'].skew())                             # skośność 
```

<h1 style="color: red;">Opis statystyczny rozkładu łącznego zmiennych</h1>

**<h2 style="color: darkorange;">Funkcja <code>groupby</code></h2>**

Funkcja `groupby` grupuje wiersze na podstawie tych samych wartości w podanej kolumnie lub kolumnach. Po niej powinna nastąpić agregacja, która wtedy zostanie wykonana osobno dla każdej grupy. Funkcja ta pozwala na przedstawianie rozkładów zmiennych ilościowych lub porządkowych względem nominalnych.

```python
df_mieszkancy.groupby('zawod')['wiek'].mean()               
```

<h3 style="color: dodgerblue;">Kilka zmiennych grupujących</h3>

```python
df_mieszkancy.groupby(['zawod', 'plec'])['wiek'].mean()   
```

<h3 style="color: dodgerblue;">Kilka zmiennych agregowanych</h3>

```python
df_mieszkancy.groupby('zawod')[['wiek', 'nr_bloku']].mean()              
```

<h3 style="color: dodgerblue;">Kilka funkcji agregujących</h3>

```python
df_mieszkancy.groupby('zawod')['wiek'].aggregate(['min', 'max', 'count'])     
```

<h3 style="color: dodgerblue;"> Kilka zmiennych grupujących/agregowanych i funkcji agregujących</h3>

```python
df_mieszkancy.groupby(['zawod', 'plec'])[['wiek', 'nr_bloku']].aggregate(['min', 'max', 'count'])#.reset_index() 
```

**<h2 style="color: darkorange;">Funkcja <code>apply</code></h2>**

Funkcja `apply` wykonuje podaną funkcję w obrębie kolumny - pozwala na połączenie właściwości funkcji `map` z funkcją agregującą.

```python
df_mieszkancy[['wiek', 'nr_bloku']].map(lambda x: x + 1)               # działania tylko na każdym elemencie osobno
```

```python
df_mieszkancy[['wiek', 'oplata_prad']].apply(lambda x: x/x.sum())      # działania na każdym elemencie z funkcją agregacyjną wywołaną w obrębie kolumny
```

Funkcję `apply` można też połączyć z funkcją `groupby`, aby funkcje agregacyjne zostały wywołane w obrębie otrzymanych przez nią podzbiorów.

```python
df_mieszkancy.groupby('zawod')['oplata_prad'].apply(lambda x: x/x.sum())
```

**<h2 style="color: darkorange;">Funkcja <code>transform</code></h2>**

Jeśli chcemy wynik funkcji `apply` z funkcją `groupby` dopisać jako kolumnę do danych potrzebujemy w zamian niej funkcji `transform`:

```python
df_mieszkancy['prad_odsetek_na_zawod'] = df_mieszkancy.groupby('zawod')['oplata_prad'].transform(lambda x: x/x.sum())  
df_mieszkancy['prad_odsetek_na_zawod']                                     
```

**<h2 style="color: darkorange;">Funkcja <code>crosstab</code></h2>**

Rozkład łączny dwóch zmiennych nominalnych można przedstawić za pomocą tabeli krzyżowej, którą wywołuje się funkcją `crosstab`:

```python
pd.crosstab(df_mieszkancy['zawod'], df_mieszkancy['plec'])
```
