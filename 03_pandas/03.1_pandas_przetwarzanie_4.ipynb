**<h2 style="color: darkorange;">8. Modyfikacja ramki</h2>**

<h3 style="color: dodgerblue;">Dodawanie / Modyfikacja kolumn</h3>

```python
df_pracownicy_cp['A'] = np.random.randint(1, 10, 12)             
df_pracownicy_cp['B'] = 99                                              # jedna wartość dla całej kolumny
df_pracownicy_cp['C'] = df_pracownicy_cp['A'] + 100                     # kolumna z przekształcenia innej
df_pracownicy_cp
```

```python
df_pracownicy_cp = df_pracownicy_cp.assign(
                   A2 = np.random.randint(1, 10, 12),                          
                   B2 = -99,
                   C2 = df_pracownicy_cp['A'] + 100,
                   D2 = lambda df: df['A2'] + 100                       # jeśli zmienna, do której się odwołujemy, została "świeżo" zdefiniowana,
                   )                                                    # potrzebujemy funkcji lambda, które "podstawia" aktualną ramkę danych
df_pracownicy_cp
```

Oba sposoby - odwołanie do kolumny i użycie funkcji `assign`, w przypadku istnienia kolumny o danej nazwie, nadpisują ją, dzięki czemu równie często służą do modyfikacji już istniejącej kolumny.

<h3 style="color: dodgerblue;">Przesuwanie kolumn</h3>

```python
df_pracownicy_cp.insert(column = 'Nazwisko',                            # nazwa
                        value  = df_pracownicy_cp.pop('Nazwisko'),      # lista wartości 
                        loc    = 0)                                     # nowa pozycja - o indeksie 0
df_pracownicy_cp
```

```python
df_pracownicy_cp.insert(column = 'Nazwisko',
                        value  = df_pracownicy_cp.pop('Nazwisko'),
                        loc    = df_pracownicy_cp.columns.get_loc('Imię') + 1)          # pozycja za kolumną Imię
df_pracownicy_cp
```

Powyższa funkcja może być użyta też do zdefiniowania nowej kolumny, gdy chcemy dodać ją w miejscu o określonym indeksie.

*Uwaga*! Funkcja `insert` działa "w miejscu" - uruchomienie automatycznie dodaje/modyfikuje kolumnę w zapisanej ramce.

<h3 style="color: dodgerblue;">Usuwanie kolumn</h3>

```python
df_pracownicy_cp.drop(columns = 'A',                                inplace = True)     # jednej kolumny
df_pracownicy_cp.drop(columns = ['B', 'C', 'A2', 'B2', 'C2', 'D2'], inplace = True)     # kilku kolumn

df_pracownicy_cp
```

<h3 style="color: dodgerblue;">Dodawanie wierszy</h3>

```python
df_pracownicy_cp.loc['BM02'] = ['Barnaba', 'Maślicki', pd.to_datetime('2002-10-05', format = '%Y-%m-%d'), 3000, 100, 'programista']   # jeden wiersz

df_pracownicy_cp = pd.concat([df_pracownicy_cp,                                         # wiele wierszy - pionowa konkatenacja ramek
                              pd.DataFrame([['Kajetan', 'Bury',    pd.to_datetime('1999-09-20', format = '%Y-%m-%d'), 2900, 200, 'sprzedawca'],
                                            ['Renata',  'Tomicka', pd.to_datetime('1990-02-09', format = '%Y-%m-%d'), 3300, 200, 'analityk']],
                                           index   = ['KB99', 'RT90'],
                                           columns = df_pracownicy_cp.columns)],
                             axis = 'rows'                                              # oś łączenia - wiersze
                             )                                                   
df_pracownicy_cp
```

<h3 style="color: dodgerblue;">Usuwanie wierszy</h3>

```python
df_pracownicy_cp.drop('BM02',           inplace = True)                                 # jednego wiersza
df_pracownicy_cp.drop(['KB99', 'RT90'], inplace = True)                                 # kilku wierszy

df_pracownicy_cp
```

**<h2 style="color: darkorange;">10. Łączenie ramek</h2>**

<h3 style="color: dodgerblue;">Konkatenacja</h3>

Przy dodawaniu wierszy podana została pozioma konkatenacja ramek. Ta sama funkcja, ale z argumentem `axis = 'columns'`, konkatenuje w poziomie, łacząc kolumny.

```python
#df_pracownicy =
pd.concat([df_pracownicy,                                                     
           pd.DataFrame({'A':np.random.randint(1, 10, 12),
                         'B':-99},
                         index = df_pracownicy.index)],
           axis = 'columns')    
```

Funkcja `concat` nie pozwala na przetwarzanie w tzw. potoku. W takiej sytuacji możemy skorzystać z funkcji `pipe` i podania w argumencie odpowiedniej funkcji.

```python
#df_pracownicy =
df_pracownicy.pipe(lambda df: pd.concat([df,
                                         pd.DataFrame({'A2':np.random.randint(1, 10, 12),
                                                       'B2':-99},
                                                      index = df_pracownicy.index)],
                                         axis = 'columns'))
```

<h3 style="color: dodgerblue;">Złączenia</h3>

**<h2 style="color: darkorange;">11. Przekształcenia ramki</h2>**

*długi format danych* - wartości zmiennej są w jednej kolumnie; każdy obserwacja ma swój wiersz

*szeroki format danych* - wartości zmiennej są w kilku kolumnach; wiersze mają po kilka obserwacji

```python
df_wide = plik_excel.parse('przypadki')
df_wide                                                    # format szeroki
```

<h3 style="color: dodgerblue;">Dane szerokie$ \rightarrow$ dane długie</h3>

```python
df_long = df_wide.melt(value_vars = [1999, 2000, 2001],    # zmienne, które mają być scalone
                       var_name   = 'Rok',                 # nazwa scalonej zmiennej
                       value_name = 'Przypadki',           # nazwa zmiennej z wartościami scalanych zmiennych
                       id_vars    = ['Kraj'])              # zmienne, które pozostawiamy       
df_long                                                    # format długi
```

<h3 style="color: dodgerblue;">Dane długie $\rightarrow$ dane szerokie</h3>

```python
df_wide_again = df_long.pivot(columns = 'Rok',             # zmienna rozdzielona na inne zmienne - jej wartości ich nazwami
                              values  = 'Przypadki',       # zmienna z rozdzielanymi wartościami
                              index   = 'Kraj')            # zmienna, która będzie indeksem
df_wide_again                                              # znów format szeroki
```

<h3 style="color: dodgerblue;">Transpozycja</h3>

*Transpozycja tabeli* to zamiana jej wierszy i kolumn ze sobą.

```python
# df_wide_again = 
df_wide_again.transpose()
```
