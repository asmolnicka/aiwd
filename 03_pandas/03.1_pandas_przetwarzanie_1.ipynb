<h3 style="color: dodgerblue;">Utworzenie serii</h3>

pandas.series mają indeksowane elementy w przeciwieństwie do list lub numpy.array.

<p style="color: darkblue;">Za pomocą <b>słownika</b></p>

````python
zmienna2 =  pd.Series({'a':'pies', 'b':'kot', 'c':'krowa'})
print(zmienna2)
````

<h3 style="color: dodgerblue;">Zdefiniowanie typu serii</h3>

pandas.series zazwyczaj składa się z elementów jednego typu:

| typ serii                | elementy serii                                                | skala zmiennej           |
| -------------------------|---------------------------------------------------------------|--------------------------|
| `str`                    | ciągi znaków                                                  | nominalna                |
| `int64`                  | liczby całkowite                                              | ilościowa dyskretna      |
| `float64`                | liczby rzeczywiste                                            | ilościowa ciągła         |
| `bool`                   | wartości logiczne                                             | ilościowa zero-jedynkowa |
| `datetime64[ns]`         | daty i/lub czasy                                              | ilościowa czasowa        |
| `category` (ordered)     | kategorie określonego zbioru z ustalonym porządkiem           | porządkowa               |
| `category` (not ordered) | kategorie określonego zbioru                                  | nominalna                |
  
Do przechowywania w serii elementów różnego typu służy typ: `object`.

`64` w nazwach typów numerycznych oznacza dokładność liczby przechowywanej w pamięci komputera, tj. na ilu bitach jest ona zapisana. \
`ns` w typie danych dotyczących dat i czasów oznacza, że są one przechowywane w postaci nanosekund.

````python
zmienna1 = pd.Series([-1, 0, 1])                      # automatycznie dopasowany typ (prawdopodobnie str, int, float, bool lub object)
zmienna2 = pd.Series([-1, 0, 1], dtype = 'float')     
zmienna3 = pd.Series([-1, 0, 1], dtype = 'str')  

print(zmienna1)
print(zmienna2)
print(zmienna3)
print(zmienna4)
````

<h3 style="color: dodgerblue;">Operacje na seriach</h3>

Pandas.series posiadają podobny (ale nie identyczny) zbiór funkcjonalności do numpy.array:

````python
zmienna1 = pd.Series([1, 2, 3])
zmienna2 = pd.Series([4, 5, 6])

print(zmienna1 + zmienna2)                             # operacje matematyczne
print(zmienna1 * 2)                                    # broadcasting
print(pd.concat([zmienna1, zmienna2]))                 # konkatenacja
print(zmienna1[zmienna1 > 1])                          # operacje logiczne i filtrowanie
zmienna1[1] = 9; print(zmienna1)                       # modyfikacja
print(np.sqrt(zmienna1))                               # funkcje matematyczne
print(zmienna1.mean())                                 # funkcje statystyczne
````

**<h2 style="color: darkorange;">2. Ramki danych - pandas.dataframe</h2>**

<h3 style="color: dodgerblue;">Utworzenie ramki danych</h3>

**<p style="color: darkblue;">Za pomocą listy</p>**

````python
tabela = pd.DataFrame([[1, 2, 'pies'],
                       [4, 5, 'kot'],
                       [7, 8, 'krowa']])
tabela
````

````python
tabela = pd.DataFrame([[1, 2, 'pies'],
                       [4, 5, 'kot'],
                       [7, 8, 'krowa']],
                      columns = ['a', 'b', 'c'],       # nazwy kolumn
                      index   = [1, 2, 3])             # indeksy wierszy
tabela
````

**<p style="color: darkblue;">Za pomocą słownika</p>**

````python
tabela = pd.DataFrame({'a': [1, 4, 7],
                       'b': [2, 5, 8],
                       'c': ['pies', 'kot', 'krowa']})
tabela
````

````python
tabela = pd.DataFrame({'a': [1, 2, 3],                 # 'nazwa_zmiennej':lista_wartości
                       'b': [4, 5, 6],
                       'c': ['pies', 'kot', 'krowa']},
                      index = ['x', 'y', 'z'])         # indeksy wierszy
tabela
````

<h3 style="color: dodgerblue;">Kopiowanie ramki danych</h3>

Tak jak w przypadku numpy.array, aby utworzyć niezależną kopię danych, która nie będzie modyfikowana wraz ze swoim oryginałem, należy użyć funkcji `copy`.

````python
df_oryginal = pd.DataFrame({'a': [1, 4, 7], 'b': [2, 5, 8], 'c': ['pies', 'kot', 'krowa']})
display(df_oryginal)

df_2 = df_oryginal
df_3 = df_oryginal.loc[:, :]
df_4 = df_oryginal.copy()                              # funkcja kopiująca
````

````python
df_oryginal['d'] = 'zoo' 

print(df_oryginal.columns)
print(df_2.columns)                                    # nie działa !
print(df_3.columns)                                    # działa (choć numpy.array[:,:] nie działa)
print(df_4.columns)                                    # działa 
````
