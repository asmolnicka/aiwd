**<h2 style="color: darkorange;">6. Filtrowanie</h2>**

<h3 style="color: dodgerblue;">Filtrowanie</h3>

Jeden warunek:

```python
print(df_pracownicy['Zarobki'] < 3100)                                      # nie "df_pracownicy[['Zarobki']]" !
display(df_pracownicy[df_pracownicy['Zarobki'] < 3100])
display(df_pracownicy.query('Zarobki < 3100'))
```

```python
print(df_pracownicy['Stanowisko'] != "analityk")
display(df_pracownicy[df_pracownicy['Stanowisko'] != "analityk"])
display(df_pracownicy.query('Stanowisko != "analityk"'))
```

Kombinacja kilku warunków:

```python
display(df_pracownicy[(df_pracownicy['Zarobki'] < 3100) & (df_pracownicy['Zarobki'] > 2900)])
display(df_pracownicy.query('Zarobki < 3100 and Zarobki > 2900'))           # & - and, | - or, ~ - not (operatory z biblioteki base)
```

<h3 style="color: dodgerblue;">Subsetting z użyciem filtrowania</h3>

```python
print(df_pracownicy[df_pracownicy['Zarobki'] < 3100]['Stanowisko'])
print(df_pracownicy['Stanowisko'][df_pracownicy['Zarobki'] < 3100])

print(df_pracownicy.loc[df_pracownicy['Zarobki'] < 3100, 'Stanowisko'])  
print(df_pracownicy.query('Zarobki < 3100').filter(['Stanowisko']))  
```

```python
display(df_pracownicy[df_pracownicy['Zarobki'] < 3100][['Data urodzenia', 'Stanowisko']])
display(df_pracownicy[['Data urodzenia', 'Stanowisko']][df_pracownicy['Zarobki'] < 3100])

display(df_pracownicy.loc[df_pracownicy['Zarobki'] < 3100, ['Data urodzenia', 'Stanowisko']])
display(df_pracownicy.query('Zarobki < 3100').filter(['Data urodzenia', 'Stanowisko']))  
```

**<h2 style="color: darkorange;">7. Modyfikacja wartości</h2>**

```python
# kopia ramki do poniższej części

df_pracownicy_cp = df_pracownicy.copy()
```

<h3 style="color: dodgerblue;">Jedna wartość</h3>

```python
print(df_pracownicy_cp.loc['MS90', 'Stanowisko'])

df_pracownicy_cp.loc['MS90', 'Stanowisko'] = 'kierownik'

print(df_pracownicy_cp.loc['MS90', 'Stanowisko'])
```

<h3 style="color: dodgerblue;">Wiele wartości</h3>

```python
print(df_pracownicy_cp.loc[['JM00', 'JB95'], 'Premia'] )

df_pracownicy_cp.loc[['JM00', 'JB95'], 'Premia'] = [100, 200]   

print(df_pracownicy_cp.loc[['JM00', 'JB95'], 'Premia'])
```

```python
print(df_pracownicy_cp.loc[['AK77', 'BS99'], 'Premia'] )

df_pracownicy_cp.loc[['AK77', 'BS99'], 'Premia'] = 250                                  # ta sama wartość dla podzbioru    

print(df_pracownicy_cp.loc[['AK77', 'BS99'], 'Premia'])
```

```python
print(df_pracownicy_cp.loc[df_pracownicy_cp['Stanowisko'] == 'sprzedawca', 'Premia'])

df_pracownicy_cp['Premia'] = df_pracownicy_cp['Premia'].mask(df_pracownicy_cp['Stanowisko'] == 'sprzedawca', 300)  
                                                                                        # ta sama wartość dla podzbioru spełniającego warunek          
print(df_pracownicy_cp.loc[df_pracownicy_cp['Stanowisko'] == 'sprzedawca', 'Premia'])
```

```python
warunek = df_pracownicy_cp['Premia'] == 200

print(df_pracownicy_cp.loc[warunek, 'Zarobki'])

df_pracownicy_cp.loc[warunek, 'Zarobki'] = df_pracownicy_cp.loc[warunek, 'Zarobki'].replace(3500, 3400)         # zmiana konkretnej/ych wartości na inne 
df_pracownicy_cp.loc[warunek, 'Zarobki'] = df_pracownicy_cp.loc[warunek, 'Zarobki'].replace({3000:3300,2800:2900}) # (o ile jest/są taka/ie w podzbiorze)
df_pracownicy_cp.loc[warunek, 'Zarobki'] = df_pracownicy_cp.loc[warunek, 'Zarobki'].replace([3400, 3200], 3000)

print(df_pracownicy_cp.loc[warunek, 'Zarobki'])
```

```python
display(df_pracownicy_cp[['Zarobki', 'Premia']])

df_pracownicy_cp.replace({'Zarobki':{3300:3500, 2900:2800},                             # zmiana konkretnej/ych wartości na inne w całej kolumnie/ach
                          'Premia' :{300:250}},
                         inplace = True)

display(df_pracownicy_cp[['Zarobki', 'Premia']])
```

`inplace = True` - działanie "w miejscu" - z uruchomieniem funkcji ramka w pamięci jest modyfikowana \
`inplace = False` - uruchomienie funkcji nie powoduje zmian w oryginalnej ramce, a tworzy zmodyfikowaną kopię ramki; \
zmiana w oryginalnej ramce wymaga wtedy przypisania wyniku do jej nazwy
